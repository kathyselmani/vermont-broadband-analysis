# 02b — FCC schema anomalies

Investigates locations listed twice (B and R) for the same provider and technology in GSO satellite and licensed fixed wireless files. Location IDs are never displayed: they belong to the FCC Fabric (CostQuest).

```python
import sys
import platform
import pandas as pd
gso = pd.read_csv(
    "../data/raw/fcc/2025-12-31/"
    "bdc_50_GSOSatellite_fixed_broadband_D25_03sep2026.csv"
)

licensed_fw = pd.read_csv(
    "../data/raw/fcc/2025-12-31/"
    "bdc_50_LicensedFixedWireless_fixed_broadband_D25_03sep2026.csv"
)
```

```python
gso_duplicates = gso[
    gso.duplicated(
        subset=["location_id", "provider_id", "technology"],
        keep=False
    )
].sort_values(
    ["location_id", "provider_id", "technology"]
)

# location_id (FCC Fabric, CostQuest IP) is hidden in displayed output
gso_duplicates.drop(columns=["location_id"]).head(20)
```

```python
licensed_duplicates = licensed_fw[
    licensed_fw.duplicated(
        subset=["location_id", "provider_id", "technology"],
        keep=False
    )
].sort_values(
    ["location_id", "provider_id", "technology"]
)

licensed_duplicates.drop(columns=["location_id"]).head(20)
```

```python
key_cols = [
    "location_id",
    "provider_id",
    "technology",
]

detail_cols = [
    "brand_name",
    "max_advertised_download_speed",
    "max_advertised_upload_speed",
    "low_latency",
    "business_residential_code",
]

gso_variation = (
    gso_duplicates
    .groupby(key_cols)[detail_cols]
    .nunique()
)

gso_variation.max()
```

```python
licensed_variation = (
    licensed_duplicates
    .groupby(key_cols)[detail_cols]
    .nunique()
)

licensed_variation.max()
```
